## ¿Cuál es el objetivo del problema?
Construir un modelo de machine learning que determine automáticamente si un paciente 
tiene enfermedad cardíaca, basándose en los resultados de sus exámenes clínicos 
(presión arterial, colesterol, frecuencia cardíaca, entre otros). La variable objetivo 
es `disease` (1 = tiene enfermedad, 0 = no tiene enfermedad).

## ¿Cómo se usará su solución?
La solución se usará como un sistema de apoyo al diagnóstico clínico. A partir de los 
resultados de exámenes de un paciente, el modelo predecirá si existe riesgo de 
enfermedad cardíaca, permitiendo a los médicos priorizar casos y tomar decisiones 
más informadas.

## ¿Cómo se debe enmarcar este problema?
- Aprendizaje supervisado: se cuenta con etiquetas (variable `disease`).
- Clasificación binaria: la salida es 1 (enfermedad) o 0 (sin enfermedad).
- Fuera de línea (batch): el modelo se entrena con datos históricos y se aplica sobre nuevos pacientes.

## ¿Cómo se debe medir el desempeño? Primera intuición.
- Recall (sensibilidad): métrica principal para minimizar falsos negativos.
- Precisión: métrica secundaria para controlar falsos positivos innecesarios.
- F1-Score: balance entre precisión y recall.
- AUC-ROC: capacidad del modelo para discriminar entre clases.

## ¿La medida de desempeño está alineada con el objetivo?
Sí. El objetivo es detectar pacientes con enfermedad cardíaca. Priorizar el Recall 
garantiza que se minimicen los casos donde un paciente enfermo sea clasificado como 
sano, lo cual está directamente alineado con el objetivo clínico.

## ¿Cuál sería el desempeño mínimo necesario?
Como línea base mínima, el modelo debe superar un clasificador trivial (ej. predecir 
siempre la clase mayoritaria). Una primera meta razonable sería:
- Recall ≥ 0.80 para la clase positiva (enfermedad = 1).
- AUC-ROC ≥ 0.75.
Estos umbrales pueden ajustarse según criterio clínico.

## ¿Cuáles son problemas parecidos? ¿Se pueden reutilizar herramientas?
Sí. Este problema es similar a otros de clasificación binaria supervisada como:
- Detección de diabetes (dataset Pima Indians): mismo tipo de variables clínicas.
- Detección de cáncer de mama (Wisconsin Breast Cancer): diagnóstico médico binario.
- Detección de spam: clasificación binaria donde un tipo de error es más costoso que el otro (Visto en clase de PLN).

Se pueden reutilizar:
- Pipelines de preprocesamiento con scikit-learn.
- Modelos como Logistic Regression, Random Forest, XGBoost.
- Técnicas de manejo de desbalance de clases (SMOTE, class_weight).

## ¿Cómo se puede resolver el problema manualmente?
Un médico cardiólogo evalúa manualmente los resultados de los exámenes del paciente 
(presión arterial, colesterol, ECG, etc.) y con base en su experiencia clínica y 
guías médicas determina si el paciente tiene riesgo de enfermedad cardíaca. Este 
proceso es costoso, no escalable y toma mucho tiempo.
